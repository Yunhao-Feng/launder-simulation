```python
import os
import torch
os.environ["ALL_PROXY"] = "socks5://127.0.0.1:20170"
os.environ["http_proxy"] = "http://127.0.0.1:20171"
os.environ["https_proxy"] = "http://127.0.0.1:20171"
from transformers import DPRContextEncoder, DPRContextEncoderTokenizerFast
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model_path = "facebook/dpr-ctx_encoder-single-nq-base"
tokenizer = DPRContextEncoderTokenizerFast.from_pretrained(model_path)
model = DPRContextEncoder.from_pretrained(model_path).to(device)
text = "The capital of France is Paris."
```

```python
tokenized = tokenizer(text, padding="max_length", truncation=True, max_length=512, return_tensors="pt")
input_ids = tokenized["input_ids"].to(device)
attention_mask = tokenized["attention_mask"].to(device)
outputs = model(input_ids=input_ids, attention_mask=attention_mask)
pooled = outputs.pooler_output #.squeeze(0).detach()
pooled.shape
```
